if an items not present in original  set.
discard() = to remove a single items that may or may not be preent in the set. this method also takes one parameter, which is the items to be removed. i
 if that items is peresnt it will remove it. it will not throw any error if it is not present.
pop()= to remove any random items from a set
clear()= to remove all items from set . the output will be an empty set
del set =delete the emtire set
