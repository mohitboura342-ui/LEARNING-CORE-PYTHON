# Operations in Pandas series


```python
import pandas as pd
s1=pd.Series([10,20,30],index=['a','b','c'])
s2=pd.Series([1,2,3],index=['a','b','c'])
result=s1+s2
result
```

```python
import pandas as pd
s1=pd.Series([10,20,30],index=['a','b','c'])  # add same index 
s2=pd.Series([1,2,3],index=['a','b','f'])
result=s1+s2
result  
```

```python
import pandas as pd
s1=pd.Series([10,20,30],index=['a','b','c'])
s2=pd.Series([1,2,3],index=['u','b','f'])
result=s1+s2
result
```

```python
# comparsion
s1=pd.Series([10,20,30])
s2=pd.Series([10,23,30])
# comparing the two Series
re=s1==s2
re
```

```python
s1[s1==s2]
```

```python
s1[(s1>5)&(s2>20)]
```

```python
# Pandas Series Index Attribute
import pandas as pd
data=pd.Series([10,20,30,40],index=['a','b','c','d'])
# Accessing the index 
print('modified series:\n',data.index)
```

```python
# modifying the index 
data.index =['w','x','y','z']
print('Modified Series:\n',data)
```

```python
# Assinging Duplicate index labels
import pandas as pd
import numpy as np
elon_ai_comp=np.array(['Tesla','Starlink','X','Solarcity','xAI'])
ser=pd.Series(elon_ai_comp)
ser.index=['a','a','c','d','e']
ser
```

```python
ser['a']
```

```python
ser.iloc[0]
```
