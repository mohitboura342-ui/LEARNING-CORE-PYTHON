## what is while loop?
#### a while loop repeats a block of code as long as a condition is true.

```python
i =1      # if loop will be upto print the looping process will continue because
while i<=5:  # 1 will  always be less than 5
    print(i) # see how the loop is running we are  increasing counting after every looping action.
    i=i+1
```

### summary : crux of making patterns
#### concept Role ###
 ### outer loop controls the number of rows (vertical dimension)
 ### inner loop controls characters per row (horizontal dimation)
 ### print(..,end='') keeps printing in the same line 
 ### print() moves to the next line after a row is done
 ### logic inner  decides what tom print:*,number,space,etc.
 ### condition (if) used for shapes like hollow boxes,diagonals etc
